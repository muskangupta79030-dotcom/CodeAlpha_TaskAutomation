```python
import re

print("📧 Email Extractor Tool")

# Step 1: Input file name
input_file = input("Enter input file name (e.g., data.txt): ")

try:
    # Step 2: Read file
    with open(input_file, "r") as file:
        content = file.read()

    # Step 3: Extract emails using regex
    emails = re.findall(r"[a-zA-Z0-9._%+-]+@[a-zA-Z0-9.-]+\.[a-zA-Z]+", content)

    # Step 4: Remove duplicates
    emails = list(set(emails))

    # Step 5: Show emails
    print("\n✅ Extracted Emails:")
    for email in emails:
        print(email)

    # Step 6: Save to file
    with open("emails.txt", "w") as file:
        for email in emails:
            file.write(email + "\n")

    print("\n💾 Emails saved to emails.txt")

except FileNotFoundError:
    print("❌ File not found! Please check the file name.")
```
